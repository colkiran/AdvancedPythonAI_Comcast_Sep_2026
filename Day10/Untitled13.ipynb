**Processing, Cleaning and Remove Duplicates**


```python
# print the count of total null values
print("Total Null Values :", df.isnull().sum())
```

```python
# if null Values in Numerical columns then
numeric_columns = [
    "Age", "Income", "Credit_Score", "Loan_Amount", "Loan_Term",
    "Existing_Loans", "Dependents" ]

for column in numeric_columns:
  df[column] = df[column].fillna(df[column].median())

# categorical Data
df["Employment"] = df["Employment"].fillna(
    df["Employment"].mode()[0] )
```

```python
# find duplicates
print("Duplicate rows :", df.duplicated().sum())

# delete the duplicate rows
df = df.drop_duplicates()
print(df.count())
```

## **Feature Engineering**

```python
# Loan to income ratio
df["Loan_Income_Ratio"] = (
    df["Loan_Amount"] / df["Income"]
)
print(df["Loan_Income_Ratio"])
```

```python
# creating a Age group
df["Age_Group"] = pd.cut(
    df["Age"], bins = [0, 30, 45, 100],
    labels = ["Young", "Middle", "Senior"]
)

print(df["Age_Group"].value_counts())
```
